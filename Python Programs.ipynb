```python
#Input 3 numbers a,b,c and print the largest of three numbers using the if statement
a=int(input("Enter the first number"))
b=int(input("Enter the second number"))
c=int(input("Enter the third number"))
# if a>b and a>c:
#     print("The largest number is",a)
large=a
if b>large:
    large=b
if c>large:
    large=c
print("The largest number is",large)
```

```python
#Input a number and print the table of the number
num=int(input("Enter the number"))
for i in range(1,11):
  print(num,"*",i,"=",num*i)
```

```python
#Input a character and check whether it is an uppercase,lowecase character or digit or any other character
ch=input("Enter the character ")
if ch>='A' and ch<='Z':
    print("The character is uppercase")
elif ch>='a' and ch<='z':
    print("The character is lowercase")
elif ch>='0' and ch<='9':
    print("The character is digit")
else:
    print("The character is special character")
```

```python
#Print sum of even or odd numbers of first n natural numbers
n=int(input("Enter the range of natural number: "))
sumeven=sumodd=0
i=0
while i<=n:
  if i%2==0:
    sumeven+=i
  else:
    sumodd+=i
  i+=1
print("The sum of even numbers is: ",sumeven)
print("The sum of odd numbers is: ",sumodd)
```

```python
#Input a number and print the next five consecutive numbers
n=int(input("Enter the number: "))
for i in range(n+1,n+6):
  print(i)
```

```python
#print each element of a list along with its index number(both forward and backward)
l1=['p','y','t','h','o','n']
length=len(l1)
for e in range(length):
  print("Forward indexing:",e,"Backward indexing:",(e-length),"Elementvalue is",l1[e])
```

```python
#print the sum of even numbers in a list
l=[1,12,6,3,8,7,13]
sum=0
for i in range(len(l)):
  if l[i]%2==0:
    sum+=l[i]
print(sum)
```
