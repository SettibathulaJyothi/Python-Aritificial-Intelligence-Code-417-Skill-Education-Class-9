#<font color=green>**Traversing a List**</font>
Traversing means visiting each element and processing it as required by a program. We use a loop to traverse a list. There are different ways of visit each element as shown below.

```python
L1=['P','Y','T','H','O','N']
for eachvalue in L1:
  print(eachvalue,end="!")
```

```python
L1=['P','Y','T','H','O','N']
for indexno in range(len(L1)):
  print(L1[indexno],end="!")
```

```python
#The above code is equivalent to:
L1=['P','Y','T','H','O','N']
print(L1[0],L1[1],L1[2],L1[3],L1[4],L1[5],sep="!",end="!")
```

```python
# To print the index number of each element in a list
L1=['P','Y','T','H','O','N']
for indexno in range(len(L1)):
  print(indexno,end="!")
```

```python
A=[1,2,3]
B=[4,5,6]
C=A+B
print(C)
```
