# <font color=green>**Some Other Functions in List**</font>

* There are some other functions also available to use with a list which are:



 **len():** This function returns the length of a list.

```python
marks = [10, 34, 42, 21, 45]
print(len(marks))
```

 **clear():** This function removes all the elements of the list in one go. It empties the list but the empty list still exists in memory which can be used later to store the values.

```python
cities = ["Delhi", "Mumbai", "Kolkata", "Chennai"]
cities.clear()
print(cities)
```

**reverse():** This function reverses the content of the list in place.

```python
alpha = ["a", "b", "c", "d", "e"]
alpha.reverse()
print(alpha)
```

**sort():** This function sorts the list in ascending or descending order. This is done in the list itself on the box for the list with values of the same data types.

```python
cities = ["Delhi", "Mumbai", "Kolkata", "Chennai"]
cities.sort()
print(cities)
cities.sort(reverse=True)
print(cities)
```

 **index():** It returns the index number of the value given in the function.

```python
cities = ["Delhi", "Mumbai", "Kolkata", "Chennai"]
print(cities.index("Mumbai"))
```

```python
print(cities.index("pune"))
```

It returns ValueError exception if the value is not found in the list

**count():** This function counts the number of occurrences of the specified value in the given list. If the value doesn't exist then the function returns 0.

```python
marks=[56,67,45,78,56,78,12]
marks.count(78)
```

```python
marks.count(10)
```
