# <font color=green>**Modifying existing values in a list**
* Lists are mutable, so data can be easily modified by overwriting a new value to an existing value in a given list by using assignment operator(=).
* Syntax to modify existing values in a list is: **list[index]=new value**
* For example,

```python
l1=[10,20,30,50]
l1[3]=100
print(l1)
```

```python
l1=[10,20,40,50]
l1[1:3]=["abc","xyz"]
print(l1)
```

# <font color=green>**Removing Elements from a List**</font>.  
* There are two different functions used to remove elements in an existing list: <font color=purple>**remove() and pop()**.</font>
* Let us learn about these functions in detail.


## <font color=purple>**The remove() function.**</font>
* The remove() function removes the first occurrence of the element with the specified value.
* It means only one value can be removed at a time, even if there are duplicate values in the list.
* If you wish to remove multiple values, then this function can be used within a loop, where it repeats itself a specific number of times.
* The syntax of remove() function is <font color=green>**list.remove(single_value)**</font>.
* For example,

```python
marks=[23,34,23,45,56,78,56]
marks.remove(23)
print(marks)
```

```python
marks=[23,34,23,45,56,78,56]
marks.remove(56)
print(marks)
```

```python
#Value Error if the value to find is not in the list
marks=[23,34,23,45,56,78,56]
marks.remove(72)
print(marks)
```

## <font color=purple>**The pop() function**</font>
* The pop() function removes an element from the list based on the index number specified in the function and returns the deleted value.
* In case no index number is given, then by default it removes the last element from the list.
* If you try to remove an index number which does not exist, then it gives an index error. For example,

```python
vowels=['a','e','i','o','u']
val=vowels.pop(2)
print(val)
```

```python
vowels=['a','e','i','o','u']
val=vowels.pop()
print(val)
```

```python
#pop index out of range
vowels=['a','e','i','o','u']
val=vowels.pop(6)
print(val)
```
