# <font color=purple>**Comparing the Lists**</font>
* The comparison of the list is done using comparison operator.
* These operators compare elements in a lexicographical order, that is alphabetically ordered.
* For example, if corresponding elements are same, it goes to the next element so on until it finds any that differ.
* For example,

```python
l1=[1,2,3]
l2=[1,2,3]
l3=[1,2,[3]]
l4=[1.0,2.0,3,0]
print(l1==l2)
print(l1>l2)
print(l1==l3)
print(l1==l4)
print([1,2,6]<[1,2,5])
print([1,2,6]!=[1,2,5])
print([1,2,6]>[1,2,5])
```

# <font color=purple>**Slicing a list**</font>
* Slicing a list means accessing a specific portion of a list.
* This can be done by using a range of index values with the start index number and end index number separated by a colon.
* We can use both forward index and backward index to specify the range of slicing in a list.
* **Some important points to consider to do slicing in a list are:**          
  + Elements from beginning to a range-<font color=purple>**list[:index] or [:-index]**</font>
  + Elements from specific index till the end-<font color=purple>**list[index:]** </font>
  + Elements within a range-<font color=purple>**list[start index:end index]**</font>
  + Elements within range using step value-<font color=purple>**list[start index:end index:step value]** </font>
  + Whole list in forward order-<font color=purple>**list[:] or list[::] or list[0:len(list)]**</font>
  + Whole list in reverse order-<font color=purple>**list[::-1]**</font>
* For example, **word = ['E', 'D', 'U', 'C', 'A', 'T', 'I', 'O', 'N']**

```python
word = ['E', 'D', 'U', 'C', 'A', 'T', 'I', 'O', 'N']
```

<font color=green>**To dispaly the whole list**</font>

```python
print(word[:])
#or
print(word[::])
#or
print(word[0:])
#or
print(word[-9:])
```

<font color=green>**To dispaly the alternate elements of the given list**</font>

```python
print(word[::2])
```

<font color=green>**To dispaly ['E','D','U'] from the above list**</font>

```python
print(word[:3])
#or
print(word[:3])
#or
print(word[-9:-6])
#or
print(word[:-6 ])
```

<font color=green>**To dispaly ['C','A','T'] from the above list**</font>

```python
print(word[3:6])
#or
print(word[-6:-3])
#or
print(word[3:-3])
```

<font color=green>**To dispaly ['O','I','T'] from the above list**</font>

```python
print(word[7:4:-1])
#or
print(word[-2:-5:-1])
#or
print(word[7:-5:-1])
#or
print(word[-2:4:-1])
```

<font color=green>**To dispaly the list in reverse order**</font>

```python
print(word[::-1])
```
