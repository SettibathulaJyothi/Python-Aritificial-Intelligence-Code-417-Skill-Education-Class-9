# <font color=blue>**Adding New Elements to List**</font>
* There are three different functions used to add elements in an existing list, which are append, extend, and insert.
* Let us learn about them in detail.


##<font color=green>**The append() Function** </font>
* The append() function appends a single element with the given value in a data type at the end of the list.
* We use this function in a loop for adding multiple elements with the single append function. The length of the list will increase by one.
* The syntax of the append function is <font color=red>**list.append(item)** </font>
  + where item can be any value like number, string, list, etc.
* For example,

```python
fruits=["apple","banana","cherries"]
#print(fruits)
fruits.append("melon")
print(fruits)
```

```python
cars=["Maruti","Ford"]
#print(cars)
cars.append("Honda")
cars.append("Hyundai")
print(cars)
```

```python
#input names of 5 friends and store in list
friends=[]
for i in range(5):
  nm=input("Enter name of friend: ")
  friends.append(nm)
print(friends)
```

```python
#adding a list at the end of another list
pets=['cat','dog','rabit']
wild=['tiger','fox']
pets.append(wild)
print(pets)
```

##<font color=green>**The extend() Function** </font>
* The extend() function is used to append multiple values or element at a time in the list
* The syntax is <font color=red>**list.extend(iterable value)**</font>
  + where iterable value can be a list.
* For Example:


```python
n=[1,2,3,4]
m=[5]
n.extend(m)
print(n)
```

```python
pets=['cat','dog','rabbit']
wild=['tiger','fox']
pets.extend(wild)
print(pets)
```

## <font color=green>**The insert() function**</font>
* The insert() functionis used to add a single value at a specific position in an existing list.
* The length of the list will increase by one.
* The syntax of the insert function is <font color=green>**list.insert(index, value)**</font>
* For example,

```python
l=[10,20,30]
#print(l)
l.insert(1,40)
print(l)
```

```python
l=[10,20,30]
#print(l)
l.insert(0,"abc")
print(l)
```

```python
l=[10,20,30]
a="xyz"
l.insert(2,a)
print(l)
```

```python
l=[10,20,30]
#print(l)
l.insert(1,["a","b"])
print(l)
```
