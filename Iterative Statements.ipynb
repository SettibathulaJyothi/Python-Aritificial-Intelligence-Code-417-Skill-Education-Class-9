# <font color=purple>**Iterative statements**

* The process of repeating a set of instructions
* There are two types of loops in python -for loop and while loop


## <font color=green>**Iterative statements**
* The for loop used to repeat a set of instructions for a fixed number of times
* for loop is also known as definite loop
* Identation of statements is must to specify the block of statements to be repeated using the for loop
* There are commonly two different ways of using the for loop:
##<font color=blue>**Using Sequence**
* A sequence of values is used over which the loop iterate
* The syntax is:\
&emsp;&emsp;for< counter variable > in < sequence >:\
&emsp;&emsp;&emsp;&emsp;Statement;\
where,
* <font color=blue>for</font> is a keyword
* <font color=blue>counter variable </font>is any identifier that keeps a count of the loop
* <font color=blue>sequence</font> is any values like integer,string,list,etc.,.
* <font color=blue>statements</font> are always indented can be single or a block

**Below are examples of for loop are as follows:**

```python
for i in [1,2,3,4]:
  print(i)
```

```python
for words in ["hello","friends","how are you?"]:
  print(words)
```

```python
for alphabets in "hello":
  print(alphabets)
```

##<font color=blue>**Using the range() function**
* The range function is an inbuilt function that is used to generate a sequence of values between the specified range.
* <font color=green>The syntax is:\
&emsp;&emsp;for<  variable > in range( < Start >,< End >,< Step >:)\
&emsp;&emsp;&emsp;&emsp;Statement;</font>\
where,
* <font color=blue>for,in,range</font> are keyword
* <font color=blue>Start,End,Step </font>are parameters of range() function and will always be integers
* <font color=blue>Start</font> is a staring value of loop
* <font color=blue>End</font> is an ending value(not iclusive) if loop
* <font color=blue>Step</font> is the number of steps taken to reach the end value
