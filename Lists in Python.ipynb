#<font color=purple>**Lists in Python**
* A list in Python is a collection of items that are ordered and mutable. It can contain elements of different data types such as strings, integers, floats, objects, and even other lists.
* Each value in a list is called an element,and each element is indexed based on its position in the list.
* The index starts at 0 for the first element, 1 for the second element, and so on.
* The number of elements in a list determines its length.
* Lists are defined by enclosing the elements within square brackets, with each element separated by a comma.
* <font color=blue>For example,\
 my_list = [1, 2, 3, "hello", 5.0]</font>
* <font color=red>Lists are mutable, meaning that you can change the elements in a list after it has been created. You can add elements to a list, modify existing elements, or remove elements from a list.

```python
my_list = [1, 2, 3, "hello", 5.0]
print(my_list)
```

##<font color=purple>**Important Features of List:**</font>

+ <font color=blue> It is mutable data type.</font >
